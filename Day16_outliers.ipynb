```python
import numpy as np
data=np.array([10,12,13,15,16,18,19,22,56,100])
Q1=np.percentile(data,25)
Q3=np.percentile(data,75)
IQR=Q3-Q1
lower=Q1-1.5*IQR
upper=Q3+1.5*IQR
print(Q1)
print(Q3)
print(IQR)
print(lower)
print(upper)
```

```python
outliers=data[(data<lower)|(data>upper)]
print(outliers)
```
