```python
#handling missing values
import pandas as pd
from sklearn.impute import SimpleImputer
data={
    "age":[18,21,None,25],
    "income":[30000,None,42000,50000]
}
df=pd.DataFrame(data)
```

```python
df
```

```python
imputer=SimpleImputer(strategy="median")
```

```python
df_imputed=imputer.fit_transform(df)
```
